```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv("cleaned_products_final.csv")
```

```python
print("=" * 60)
print("MOTORPH PRODUCT INVENTORY ANALYSIS")
print("=" * 60)
```

```python
# Total number of products
total_products = len(df)
print(f"\Total Number of Products: {total_products}")
```

```python
# Product types
print("\nPRODUCTS BY PRODUCT TYPE")
print("-" * 40)
print(df["Product Type"].value_counts())
```

```python
print("\nPRODUCTS BY PRODUCT TYPE with Percentile")
product_count = df["Product Type"].value_counts()

for product, count in product_count.items():
    percentage = (count / total_products) * 100
    print(f"{product}: {count} products ({percentage:.0f}%)")
```

```python
# Price statistics
print("\nPRICE STATISTICS")
print("-" * 40)
print(f"Average Price: PHP {df['Unit Price'].mean():,.2f}")
print(f"Minimum Price: PHP {df['Unit Price'].min():,.2f}")
print(f"Maximum Price: PHP {df['Unit Price'].max():,.2f}")
print(f"Median Price: PHP {df['Unit Price'].median():,.2f}")
```

```python
# Total inventory cost
print(f"\nTotal Inventory Cost: PHP {df['Unit Price'].sum():,.2f}")
```

```python
# Manufacturing years
print("\nMANUFACTURING YEAR DISTRIBUTION")
print("-" * 40)
print(df["Date of Manufacturing"].value_counts().sort_index())
```

```python
# Acquisition years
print("\nACQUISITION YEAR DISTRIBUTION")
print("-" * 40)
print(df["Date of Acquisition"].value_counts().sort_index())
```
